## Lectura y escritura de archivos

    f = open("archivo.txt", "w")
    f.write('Hola mundo\n')
    f.close()

tenemos los siguientes modos:
- 'r' : read, lectura. Abre el archivo en modo lectura.
- 'w' : write, escritura. Abre el archivo en modo escritura, sobreescribiendo el contenido si el archivo ya existe.
- 'a' : append, añadir. Se diferencia del modo 'w' en que no se sobreescribe el contenido del archivo, sino que se comienza a escribir al final del archivo.
- 'b' : binary, binario.
- '+' : permite lectura y escritura simultáneas.

La forma recomendada de trabajar con archivos es con `with`, que cierra el archivo automáticamente al salir del bloque (incluso si ocurre un error):

    with open("archivo.txt", "r") as f:
        completo = f.read()

```python
f = open("archivo.txt", "r")
print(f)
```

```python
f = open("archivo.txt",encoding='cp437')
print(f)
f = open("archivo.txt",encoding='utf8')
print(f)
```

```python
f = open("archivo.txt", "r")
completo = f.read()
print(completo)
f.close()
```

```python
with open("archivo.txt") as f:
    completo = f.read()
print(completo)
```

```python
f = open("archivo.txt", "r")
parte = f.read(8)
print(parte)
f.close()
```

```python
f = open("archivo.txt", "r")
while True:
    linea = f.readline()
    if not linea: 
        break
    print(linea)
f.close()
```

```python
archivo="archivo.txt"
text=open(archivo).readlines()
for line in text:
    print(line.split())
```

## Manejando excepciones

```python
archivo="archivo.txt"
try:
    f = open(archivo, 'r')
except FileNotFoundError:
    print('no pude abrir', archivo)
else:
    print(archivo, 'tiene', len(f.readlines()), 'lineas')
    f.close()

archivo="no_existe.txt"
try:
    f = open(archivo, 'r')
except FileNotFoundError:
    print('no pude abrir', archivo)
else:
    print(archivo, 'tiene', len(f.readlines()), 'lineas')
    f.close()
```

## Format
La forma más sencilla de mostrar algo en la salida estándar es mediante el uso de la sentencia print, la cual permite también utilizar técnicas avanzadas de formateo, de forma similar al sprintf de C.
Los especificadores más sencillos están formados por el símbolo % seguido de una letra que indica el tipo con el que formatear el valor. Por ejemplo:

Especificador | Formato
:--------: | :-------:
%s |Cadena
%d |Entero
%o |Octal
%x |Hexadecimal
%f |Real


```python
print ("%10s mundo" % "Hola")
```

```python
print ("%-10s mundo" % "Hola")
```

```python
pi=3.141592653589793115997963
print("%.24f" % pi)
print("%.6f" % pi)
print("%.4f" % pi,"%.6f" % pi)
```

```python
print ("%.4s" % "hola mundo")
```

```python
n=123.456789
cadena='{0:12.6f}  '.format(n)
cadena=cadena+'{0:12.6f}  '.format(n)
cadena=cadena+"\n"
n=9876.54321
cadena=cadena+'{0:12.6f}  '.format(n)
cadena=cadena+'{0:12.6f}  '.format(n)
print(cadena)
```

## f-strings
Desde Python 3.6 la forma más habitual de formatear es con *f-strings*: cadenas con el prefijo `f` en las que las expresiones entre llaves se sustituyen por su valor, con la misma sintaxis de formato que `.format()`.

```python
pi = 3.141592653589793
n = 123.456789
print(f"pi vale {pi:.4f}")
print(f"{n:12.6f}  {n:12.6f}")
print(f"{'Hola':>10} mundo")
```
